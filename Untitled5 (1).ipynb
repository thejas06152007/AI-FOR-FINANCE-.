```python
# @title 🐍 Step 3: Run Flask Application
from flask import Flask, render_template, request, jsonify
from flask_cors import CORS
from google import genai
from google.colab import userdata
import threading

app = Flask(__name__)
CORS(app)

# Initialize Google Gemini Client securely
try:
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
    client = genai.Client(api_key=GOOGLE_API_KEY)
except Exception:
    client = None
    print("⚠️ Warning: GOOGLE_API_KEY not found in secrets. Chat will operate in fallback mode.")

@app.route('/')
def index():
    return render_template('index.html')

@app.route('/api/chat', methods=['POST'])
def chat():
    user_message = request.json.get('message', '')

    if not client:
        return jsonify({"reply": "Fallback Mode: It looks like the GOOGLE_API_KEY secret is not set yet in your Colab Secrets (the key icon). Please set it to enable full AI reasoning."})

    try:
        system_instruction = (
            "You are FinancAI, a professional and extremely helpful personal financial assistant. "
            "Provide smart, helpful, and safe financial guidance on budgeting, loans, savings, and investments. "
            "Do not give legal/regulatory financial advice, but offer structured educational suggestions."
        )
        response = client.models.generate_content(
            model='gemini-2.5-flash-lite',
            contents=user_message,
            config={'system_instruction': system_instruction}
        )
        return jsonify({"reply": response.text})
    except Exception as e:
        return jsonify({"reply": f"Error calling Gemini: {str(e)}"})

def run_app():
    app.run(host='0.0.0.0', port=5000)

# Start server on a background thread
threading.Thread(target=run_app, daemon=True).start()
print("🚀 Flask Server running in the background on port 5000!")
```
